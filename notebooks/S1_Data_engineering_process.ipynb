# Data Engineering Process

Limpieza, validación y transformación de datos de desempeño laboral con Python.


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import missingno as msno
```

```python
# Cargar la base original desde la carpeta data
df = pd.read_csv("../data/Performance_workers.csv")
print(df.shape)
```

```python
# Inspeccionar la forma y cinco registros aleatorios
print(df.shape)
df.sample(5)
```

```python
#ver si hay datos vacios
msno.matrix(df)
```

```python
#Que tantos datos vacios hay por columna
msno.bar(df)
```

```python
#Determinar el valor exacto de datos vacíos por columna
df.isna().sum()
```

```python
#Eliminar columnas que no sirven
#Axis igual a 1 elimina columna axis=o elimina fila
#drop para columnas y dropna filas vacias
df.drop(['Unnamed: 0'], axis=1, inplace=True)
print(df.shape)
df.sample(5)
```

```python
#1. Primer paso eliminar los vacios de la columna con menor cantidad de datos vacios
df.dropna(subset=['Nombre'], axis=0, inplace=True)
print(df.shape)
df.sample(5)
```

```python
#Imputar la cuidad con la palabra desconocido
df['Ciudad']=df['Ciudad'].fillna('Desconocido')
print(df.shape)
print(df.shape)
df.sample(5)
```

```python
df.isna().sum()
```

```python
#Cual es la categoria de ciudad que más se repite
df['Ciudad'].value_counts()
```

```python
#reemplazar los vacios de comentarios por "No especificado"
df['Comentarios']=df['Comentarios'].fillna('No especificado')
print(df.shape)
print(df.shape)
df.sample(5)
```

```python
#Verificar que funciono el fillna
df.isna().sum()
#Imputar con mediana si hay Outliers
#Imputar con promedio media no outliers
#Imputar moda datos categoricos
```

```python
#Cual es el comentario que más se repite
df['Comentarios'].value_counts()
```

```python
#Exploramos la presencia de outliers : Edad
#estamos usando pandas aquí
df['Edad'].plot(kind='box')
#COMO NO HAY OULIERS SE IMPUTA CON EL PROMEDIO O MEDIA
```

```python
#Exploramos la presencia de outliers : Edad
df['Puntuacion'].plot(kind='box')
```

```python
#reemplazar los vacios de Edad por el promedioo media
df['Edad']=df['Edad'].fillna(df['Edad'].mean())
print(df.shape)
df.sample(5)
```

```python
#reemplazar los vacios de puntuación por el promedioo media
df['Puntuacion']=df['Puntuacion'].fillna(df['Puntuacion'].mean())
print(df.shape)
df.sample(5)
```

```python
#Validamos que los datos esten limpios
df.isna().sum()
```

```python
#Validar si el tipo de datos es correcto por columnas
df.dtypes
```

```python
#Corregir tipo de datos: float a entero
df['Edad']=df['Edad'].astype(int)
df.dtypes

```

```python
#Corregir tipo de datos: object a category
df[['Nombre','Ciudad','Comentarios']] = df[['Nombre','Ciudad','Comentarios']].astype('category')

df.dtypes
```

```python
df['Nombre'].value_counts() #revisar si hay espacios en blanco
```

```python
#Funcion para eliminar espacio en blanco se llama STRIP
df['Nombre']= df['Nombre'].str.strip()
df['Nombre'].value_counts()
```

## Eliminación de duplicados


```python
#Guardar tamaño inicial de la base
tam_ini = df.shape[0]
tam_ini
```

```python
#Borramos los duplicados
df = df.drop_duplicates(subset=df.columns)
print(df.shape)
print('Eliminamos {} registros'.format(tam_ini-df.shape[0]))
df.sample(5)
```

## Transformación de datos


```python
#Obtener el rendimiento según: Puntuación/ max(puntuación)*100 [Funcion landa es más rapido]
df['Rendimiento'] = df['Puntuacion'].apply(
    lambda x: (x / df['Puntuacion'].max()) * 100
)

print(df.shape)
df.head(5)
```

```python
#Calcular el índice de juventud =[edad/max(rendimiento))*100]
df['Juventud']= (df['Edad']/ df['Rendimiento'].max())*100
print(df.shape)
df.head(5)
```

```python
# Guardar la base limpia en formato CSV
df.to_csv("../outputs/Clean_data_PerformanceWorkers.csv", index=False)
```

```python
# Guardar la base limpia en formato Excel
df.to_excel("../outputs/Clean_data_PerformanceWorkers.xlsx", index=False)
```
